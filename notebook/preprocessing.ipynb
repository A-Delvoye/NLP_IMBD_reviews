### 1. Convert to lowercase
### 2.a. Remove punctuation (perso)

```python
import re
test = test.lower()
print(f"texte initial : {test}")
test_no_symbols = re.sub(r'[^\w\s]', ' ', test) # all except letters, numbers, spaces
print(f"texte sans symboles : {test_no_symbols}")
```

### 2.b Remove punctuation

```python
import string

def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)
test_no_symbols = remove_punctuation(test)
```

```python
test_clean = re.sub(r'\s+', ' ', test_no_symbols).strip()
print("Texte final nettoyé :", test_clean)
```

### 3.b. Convert numbers into letters

```python
import inflect
p = inflect.engine()

# test_clean = "4 bananes enragées font 1 troupeau de bananes"

def convert_number(text):
    temp_str = text.split()
    new_string = []

    for word in temp_str:
        if word.isdigit():
            temp = p.number_to_words(word)
            new_string.append(temp)

        else:
            new_string.append(word)

    temp_str = ' '.join(new_string)
    return temp_str

test_no_numbers = convert_number(test_clean)
print(test_no_numbers)
```

### 4. Removing stopwords

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('stopwords')


def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word not in stop_words]
    return filtered_text

no_stopwords = remove_stopwords(test_no_numbers)
print(no_stopwords)
```

```python
len(no_stopwords)
print(no_stopwords)
```

### 5.a. Stemming

```python
from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()

def stem_words(text):
    stems = [stemmer.stem(word) for word in no_stopwords]
    return stems

stem = stem_words(no_stopwords)
print(stem)
print(len(stem))
processed_data = ' '.join(stem)
print(processed_data)
```

### 5.b. Lemmatization

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()

def lemma_words(text):
    word_tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(word) for word in word_tokens]
    return lemmas
  
input_str = "data science uses scientific methods algorithms and many types of processes"
lemma_words(input_str)
```

# Sentiment Analysis

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import emoji

def analyze_sentiment_with_emojis(text):
    # Convert emojis to text
    text = emoji.demojize(text)
    
    # Analyze sentiment
    analyzer = SentimentIntensityAnalyzer()
    sentiment = analyzer.polarity_scores(text)
    
    return sentiment

# Example usage
text_with_emojis = "I'm so excited! 😊 all my trouble seem so far away"
sentiment = analyze_sentiment_with_emojis(text_with_emojis)
print("Sentiment scores:", sentiment)
```
